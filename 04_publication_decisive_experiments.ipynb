# 04. Publication decisive experiments

## tl;dr

The completed development-validation study supports a refinement branch, but it does not yet establish that the directional adaptive stencil is the reason for the gain.

Across matched validation cases, the full model reduces scaled L1 by about **1.82%**, pressure relative L2 by **8.35%**, velocity relative L2 by **8.12%**, and near-airfoil velocity error by **13.06%** relative to the no-refinement model. The wake-velocity interval crosses zero. Alignment and learned stencil weights remain inconclusive or adverse.

This notebook resolves the remaining publication questions in a fixed order:

1. Generate matched CFD field figures and inference-cost measurements for the completed ablations.
2. Train the conventional CNN refiner control across three seeds.
3. Train DFP, residual U-Net, and FNO baselines under the same 10,000-step validation protocol.
4. Compare completed models using paired case-level intervals and regional errors.
5. Freeze the final architecture before a separate official-test evaluation.

The official test set is not accessed here. Existing runs are read-only inputs. New runs remain under `runs/study_04/matched_validation_10k_3seeds`.

**Saved-output status:** the delivered notebook is executed in snapshot mode to verify its portable analysis path. Drive record audits, checkpoint figures, inference timing, and GPU training are intentionally not presented as completed until you run the corresponding action in Colab.

## 3. Existing-model figures and inference costs

This action produces nine matched validation figures, three selected cases times three predicted fields, and inference measurements for batch sizes 1 and 10. The cases are selected from the no-refinement model's median, 95th-percentile, and worst validation errors. They are diagnostics, not independent test examples.

The timing protocol uses five warmup passes and 30 measured resident-tensor FP32 forward passes. It excludes preprocessing and host-to-device transfer. Use the same GPU for every model and report the GPU name.


```python
core = plan = None

if MODE == "drive":
    core = stage.load_core(CORE_PATH, audit["core_fingerprint"])

    def expected_next_configs():
        '''Rebuild only the immutable experiment configurations.'''
        template = stage.copy.deepcopy(
            audit["identities"][("proposed", 42)]["config"]
        )
        matched_settings = [
            "max_steps", "batch_size", "validation_every", "lr", "betas",
            "weight_decay", "grad_clip", "num_workers", "loss", "use_sdf",
            "precision", "selection_metric",
        ]
        configurations = {}

        for experiment_name in ["dfp_original", "modern_unet", "fno"]:
            configuration = stage.copy.deepcopy(
                core.experiment_configs("screen")[experiment_name]
            )
            configuration.update({
                key: stage.copy.deepcopy(template[key])
                for key in matched_settings
            })
            configuration["profile"] = "matched_validation"
            configurations[experiment_name] = configuration

        control = stage.copy.deepcopy(template)
        control.update(
            model="conv_refiner_control",
            profile="matched_validation",
        )
        configurations["conv_refiner_control"] = control
        return configurations


    def load_or_prepare_saved_plan():
        '''Reuse a compatible saved plan across changing Colab runtimes.'''
        plan_path = NEXT_ROOT / "next_stage_plan.json"

        if not plan_path.is_file():
            return stage.prepare_plan(core, audit, CORE_PATH, NEXT_ROOT)

        saved_plan = stage.read_json(plan_path)
        plan_without_id = {
            key: value
            for key, value in saved_plan.items()
            if key != "plan_id"
        }
        stage.require(
            saved_plan.get("plan_id") == stage.json_hash(plan_without_id),
            "Saved next-stage plan hash is invalid.",
        )

        expected_source_runs = {
            f"{name}_seed{seed}": identity["run_id"]
            for (name, seed), identity in audit["identities"].items()
        }
        immutable_checks = {
            "schema": stage.VERSION,
            "core_fingerprint": audit["core_fingerprint"],
            "core_source_sha256": stage.file_hash(CORE_PATH),
            "runner_sha256": stage.file_hash(stage.__file__),
            "dataset_id": audit["metadata"]["dataset_id"],
            "split_id": audit["split"]["split_id"],
            "configs": expected_next_configs(),
            "seeds": stage.SEEDS,
            "source_run_ids": expected_source_runs,
        }

        for field, expected_value in immutable_checks.items():
            stage.require(
                saved_plan.get(field) == expected_value,
                f"Saved next-stage plan has an incompatible {field}.",
            )

        current_environment = {
            "python": stage.platform.python_version(),
            "torch": str(core.torch.__version__),
            "numpy": np.__version__,
        }
        saved_environment = saved_plan.get("source_software_environment", {})
        if saved_environment != current_environment:
            print("Reusing the verified saved study plan across a runtime change.")
            print("Saved environment:", saved_environment)
            print("Current environment:", current_environment)
            print("Each training session is recorded separately for reproducibility.")
        else:
            print("Reusing the verified saved study plan.")

        return saved_plan


    plan = load_or_prepare_saved_plan()

if MODE == "drive" and EFFECTIVE_ACTION == "diagnostics":
    diagnostics_root = REPORT_ROOT / "existing_model_diagnostics"
    diagnostics_root.mkdir(parents=True, exist_ok=True)
    timing = stage.validation_diagnostics(
        core,
        audit,
        PREPARED_ROOT,
        STUDY_ROOT,
        diagnostics_root,
        make_figures=True,
        measure_inference=True,
    )
    display(timing)
    print("Diagnostics completed. Next action: train_control.")
else:
    print("Diagnostics not requested in this execution.")
```

## 5. Train matched-budget baselines

The runner completes one baseline across all three seeds before moving to the next one. Priority is DFP, residual U-Net, then FNO. This yields an interpretable three-seed comparison as early as possible.

The comparison is budget matched and uses the same split, step budget, validation interval, checkpoint-selection metric, and seeds. It is not a claim that every baseline received exhaustive architecture-specific hyperparameter tuning, so that limitation must be reported.


```python
if MODE == "drive" and plan is not None and EFFECTIVE_ACTION == "train_baselines":
    baseline_progress = stage.progress_table(NEXT_ROOT, plan)
    baseline_name = None

    for candidate_name in BASELINE_PRIORITY:
        candidate_rows = baseline_progress[baseline_progress.variant == candidate_name]
        if not (candidate_rows.status == "completed").all():
            baseline_name = candidate_name
            break

    if baseline_name is None:
        print("All matched-budget baselines are complete.")
    else:
        print("Active baseline:", baseline_name)
        baseline_progress = stage.run_next_chunk(
            core,
            PREPARED_ROOT,
            audit["split"],
            NEXT_ROOT,
            plan,
            [baseline_name],
            chunk_steps=CHUNK_STEPS,
            session_minutes=SESSION_MINUTES,
        )
        display(baseline_progress[baseline_progress.variant.isin(BASELINE_PRIORITY)])
        print("Rerun with ACTION='train_baselines' to continue automatically.")
else:
    print("Baseline training not requested in this execution.")
```

Only configurations with all three completed seeds enter the comparison. Partial configurations remain visible in the progress table but cannot support a three-seed model claim.

The key control comparisons are against both the full model and the Cartesian-stencil variant, because the latter currently has the lowest mean global validation errors. Near ties should be reported as ties unless the paired intervals support a stable difference.


```python
comparison_intervals = comparison_regions = combined_runs = None

if MODE == "drive" and plan is not None:
    next_progress = stage.progress_table(NEXT_ROOT, plan)
    display(next_progress)

    complete_names = [
        name
        for name in plan["configs"]
        if (next_progress[next_progress.variant == name].status == "completed").all()
    ]

    if complete_names:
        new_audit = stage.audit_study(
            PREPARED_ROOT,
            SPLIT_PATH,
            NEXT_ROOT,
            names=complete_names,
            seeds=plan["seeds"],
            expected_configs=plan["configs"],
        )

        combined_cases = pd.concat([audit["cases"], new_audit["cases"]], ignore_index=True)
        combined_runs = pd.concat([audit["runs"], new_audit["runs"]], ignore_index=True)

        comparisons = [
            (name, candidate)
            for name in complete_names
            for candidate in ["proposed", "no_alignment"]
        ]
        comparison_intervals, comparison_regions = stage.regional_analysis(
            combined_cases,
            comparisons,
            BOOTSTRAP_REPETITIONS,
        )

        stage.write_csv(REPORT_ROOT / "matched_budget_completed_runs.csv", combined_runs)
        stage.write_csv(REPORT_ROOT / "matched_budget_intervals.csv", comparison_intervals)
        stage.write_csv(REPORT_ROOT / "matched_budget_regional_summary.csv", comparison_regions)

        display(combined_runs[[
            "variant", "seed", "steps", "parameters", "validation_L1",
            "validation_pressure_L2", "validation_velocity_L2",
        ]].sort_values(["variant", "seed"]).round(6))

        display(comparison_intervals[
            comparison_intervals.metric.isin(stage.METRICS.values())
        ][[
            "reference", "candidate", "metric", "reduction_percent",
            "absolute_ci_low", "absolute_ci_high", "valid_pairs",
        ]].round(6))
    else:
        print("No new configuration has all three completed seeds yet.")
else:
    print("Matched-budget comparisons require live completed records.")
```

## 7. Publication gate and final-test handoff

Do not access the official test set until all required validation-stage decisions are complete.

| Gate | Required evidence |
|---|---|
| Mechanism | Three completed CNN-control seeds and paired case-level comparison |
| Baseline superiority | Three completed seeds for DFP, U-Net, and FNO at the matched budget |
| Qualitative credibility | Shared-scale CFD target, prediction, and error figures, including hard cases |
| Efficiency | Inference latency, memory, parameter count, and clearly defined timing scope |
| Architecture freeze | One selected model, fixed selection metric, budget, preprocessing, and checkpoint rule |
| Independent generalization | Official unseen-airfoil test, evaluated once after freezing |

Possible outcomes must be reported honestly:

* If the CNN control matches the stencil models, claim a geometry-conditioned iterative refinement framework, not a demonstrated benefit of directional adaptive stencils.
* If the Cartesian stencil remains best, select or discuss the simpler variant and report that alignment was not beneficial.
* If matched baselines are competitive, frame the work as a component and error-localization study rather than a state-of-the-art accuracy claim.
* If the selected model wins consistently, proceed to a separate official-test notebook with paired airfoil-level uncertainty and no further architecture selection.
